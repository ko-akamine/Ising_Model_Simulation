```python
import numpy as np
import matplotlib.pyplot as plt

# --- 1. パラメgit add ising.ipynbータ設定 ---
N = 100  # 格子のサイズ (100 x 100)

# --- 2. 物理的な状態の生成 ---
# スピン変数は +1 (上向き) か -1 (下向き) のどちらかをとる
# np.random.choice で、50%の確率でランダムに配置する
# 物理的意味: これは「温度無限大 (T=∞)」の極限状態に対応します
spins = np.random.choice([-1, 1], size=(N, N))

# --- 3. 可視化 (スナップショット) ---
plt.figure(figsize=(6, 6))
plt.imshow(spins, cmap='binary', interpolation='nearest')
plt.title(f"Initial State (T -> infinity): {N}x{N}")
plt.colorbar(label="Spin Direction")
plt.show()
```

```python
# --- 4. メトロポリス法の実装（物理エンジンの核心） ---

def metropolis_step(spins, T):
    """
    スピン配位を1ステップ更新する関数
    T: 温度 (Temperature)
    """
    N = spins.shape[0]
    
    # 1回のステップで、N*N回（全粒子数分）の更新を試みるのが通例
    for _ in range(N * N):
        # 1. ランダムに1つのスピンを選ぶ
        i = np.random.randint(0, N)
        j = np.random.randint(0, N)
        s = spins[i, j]
        
        # 2. エネルギー変化 dE を計算する (周期境界条件)
        # 隣り合う4つのスピンの和をとる
        neighbor_sum = (
            spins[(i+1)%N, j] + # 下
            spins[(i-1)%N, j] + # 上
            spins[i, (j+1)%N] + # 右
            spins[i, (j-1)%N]   # 左
        )
        
        # dE = 2 * s * (隣の和) 
        # (J=1 としている。もしフリップしたらエネルギーはどう変わるか？)
        dE = 2 * s * neighbor_sum
        
        # 3. 判定 (Metropolis Criterion)
        # エネルギーが下がる(dE < 0)なら必ず裏返す
        # 上がる場合でも、確率 exp(-dE/T) で裏返す（熱ゆらぎ）
        if dE < 0:
            spins[i, j] *= -1 # フリップ！
        elif np.random.rand() < np.exp(-dE / T):
            spins[i, j] *= -1 # 確率的にフリップ！

# --- 5. シミュレーション実行 ---

# 温度を設定（臨界温度 Tc ≒ 2.27 より低くしてみる）
Temperature = 1.5 
Time_Steps = 100  # 時間を進める回数

print(f"Simulation Start: T={Temperature} (Low Temp)")

# 時間発展させる
for t in range(Time_Steps):
    metropolis_step(spins, Temperature)

print("Simulation Finished!")

# --- 6. 結果の表示 ---
plt.figure(figsize=(6, 6))
plt.imshow(spins, cmap='binary', interpolation='nearest')
plt.title(f"After {Time_Steps} Steps at T={Temperature}")
plt.colorbar(label="Spin Direction")
plt.show()
```

```python
# --- 5. 相転移の実験（Temperature Sweep） ---

# 実験設定
N_sim = 32          # 計算を早くするためサイズを小さくします
T_steps = 50        # 温度を何段階変えるか
T_max, T_min = 4.0, 1.0 # 温度の範囲
steps_per_T = 1000  # 各温度でのシミュレーション回数

# 温度の配列を用意 (4.0 から 1.0 へ下げる)
temperatures = np.linspace(T_max, T_min, T_steps)
magnetizations = [] # 結果を記録するリスト

print(f"Experiment Start: Sweeping T from {T_max} to {T_min}...")

# --- 実験ループ ---
for T in temperatures:
    # 1. 毎回ランダムな状態からスタート（独立試行）
    spins = np.random.choice([-1, 1], size=(N_sim, N_sim))
    
    # 2. 時間発展（熱平衡状態になるまで待つ）
    # 平均磁化を測るために、後半のステップだけを使うのが正確ですが、
    # 今回は簡易的に最終状態の磁化を使います。
    for _ in range(steps_per_T):
        metropolis_step(spins, T)
    
    # 3. 磁化の測定
    # 全スピンの和をとって、格子点数(N*N)で割る。絶対値をとる。
    m = np.abs(np.sum(spins)) / (N_sim * N_sim)
    magnetizations.append(m)
    
    # 進捗表示 (10回に1回表示)
    if T == temperatures[0] or T == temperatures[-1] or len(magnetizations) % 10 == 0:
        print(f"T = {T:.2f}, Magnetization = {m:.3f}")

# --- 6. 結果の可視化（グラフ化） ---
plt.figure(figsize=(8, 5))
plt.plot(temperatures, magnetizations, 'o-', color='blue', markersize=4, alpha=0.7)
plt.axvline(x=2.269, color='red', linestyle='--', label='Critical Temp (Tc ≈ 2.27)')
plt.xlabel("Temperature (T)")
plt.ylabel("Magnetization |M|")
plt.title("Phase Transition of 2D Ising Model")
plt.grid(True)
plt.legend()
plt.gca().invert_xaxis() # 温度が高い方(左)から低い方(右)へ見るのが通例
plt.show()
```
